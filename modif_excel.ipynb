```python
from datetime import time
import pandas as pd
import numpy as np

def convert_to_time(value):
    if pd.isna(value) or value == '' or value is None:
        return None
    try:
        str_val = str(value).strip()
        if '.' in str_val:
            jam, menit = str_val.split('.')
            jam = int(jam)
            menit = int(menit.ljust(2, '0')[:2])
        else:
            jam = int(str_val)
            menit = 0
        return time(jam, menit)
    except:
        return None

# Terapkan ke kolom
df["Mulai Pengerjaan"] = df["Mulai Pengerjaan"].apply(convert_to_time)
df["Selesai Pengerjaan"] = df["Selesai Pengerjaan"].apply(convert_to_time)

# Tampilkan hasil
df
```

```python
def hitung_kerusakan(teks):
    if not teks or teks == "":
        return 0
    if pd.isna(teks) or teks == '' or teks is None:
        return 0
    
    # Hitung jumlah item berdasarkan koma
    jumlah_kerusakan = teks.count(',') + 1

    # Jika mengandung "cek rutin", kurangi 1
    if 'cek rutin' in teks.lower():
        jumlah_kerusakan -= 1

    return jumlah_kerusakan
```

```python
df['Total Kerusakan'] = df["Jenis Kerusakan"].apply(hitung_kerusakan)
```

```python
import pandas as pd
import numpy as np

# Contoh pembacaan data dari file Excel (sesuaikan dengan sumber data yang digunakan)
df = pd.read_excel("HM_Harian.xlsx")

# Mengubah kolom kecuali 'Tanggal' menjadi numerik jika memungkinkan, jika tidak biarkan dalam bentuk aslinya
def convert_to_numeric(value):
    try:
        return float(value)
    except ValueError:
        return value  # Jika bukan angka, biarkan tetap dalam bentuk aslinya

df.iloc[:, 1:] = df.iloc[:, 1:].applymap(convert_to_numeric)

# Menghitung penggunaan forklift tiap shift dengan mempertahankan nilai non-numerik
usage_df = df.copy()

num_forklifts = 26  # Total forklift
shifts_per_forklift = 3  # Shift per forklift
total_columns = num_forklifts * shifts_per_forklift

cols = df.columns[1:total_columns + 1]  # Semua kolom kecuali 'Tanggal'

for i in range(num_forklifts):  # Loop per forklift
    shift_1_idx = i * shifts_per_forklift
    shift_2_idx = shift_1_idx + 1
    shift_3_idx = shift_1_idx + 2
    
    shift_1_col = cols[shift_1_idx]
    shift_2_col = cols[shift_2_idx]
    shift_3_col = cols[shift_3_idx]
    
    # Shift 1 = Shift 2 - Shift 1 (jika keduanya angka, jika tidak salin saja)
    usage_df[shift_1_col] = df.apply(lambda row: row[shift_2_col] - row[shift_1_col] 
                                     if isinstance(row[shift_2_col], (int, float)) and isinstance(row[shift_1_col], (int, float))
                                     else row[shift_1_col], axis=1)
    
    # Shift 2 = Shift 3 - Shift 2
    usage_df[shift_2_col] = df.apply(lambda row: row[shift_3_col] - row[shift_2_col] 
                                     if isinstance(row[shift_3_col], (int, float)) and isinstance(row[shift_2_col], (int, float))
                                     else row[shift_2_col], axis=1)
    
    # Shift 3 = Shift 1 (hari berikutnya) - Shift 3
    usage_df[shift_3_col] = df.apply(lambda row: row[shift_3_col] if isinstance(row[shift_3_col], str) else 
                                     (df[shift_1_col].shift(-1).iloc[row.name] - row[shift_3_col] 
                                      if row.name < len(df) - 1 and isinstance(df[shift_1_col].shift(-1).iloc[row.name], (int, float)) 
                                      and isinstance(row[shift_3_col], (int, float)) else row[shift_3_col]), axis=1)

# Menyimpan hasil ke file baru
usage_df.to_excel("penggunaan_forklift.xlsx", index=False)

print("Perhitungan selesai. Data disimpan di 'penggunaan_forklift.xlsx'")
```

```python
import pandas as pd
import numpy as np

# Baca data dari file Excel (sesuaikan nama file jika perlu)
df = pd.read_excel("HM_Harian.xlsx")

# Fungsi untuk mencoba mengonversi ke float, jika tidak berhasil maka biarkan nilai aslinya
def convert_to_numeric(value):
    try:
        return float(value)
    except (ValueError, TypeError):
        return value

# Ubah semua kolom kecuali 'Tanggal'
df.iloc[:, 1:] = df.iloc[:, 1:].applymap(convert_to_numeric)

# DataFrame untuk hasil penggunaan
usage_df = df.copy()

num_forklifts = 26       # Total forklift
shifts_per_forklift = 3  # Shift per forklift
total_columns = num_forklifts * shifts_per_forklift

# Kolom-kolom yang akan dihitung (semua kolom kecuali Tanggal)
cols = df.columns[1:total_columns + 1]

# Proses per forklift
for i in range(num_forklifts):
    # Dapatkan daftar kolom untuk forklift ke-i
    forklift_cols = [cols[i * shifts_per_forklift + j] for j in range(shifts_per_forklift)]
    
    # Kumpulkan posisi (baris, kolom) untuk forklift ini secara berurutan
    positions = []
    for r in range(len(df)):
        for col in forklift_cols:
            positions.append((r, col))
    
    # Untuk setiap posisi, jika nilai adalah angka, cari nilai angka berikutnya dalam urutan positions
    for index, (r, col) in enumerate(positions):
        current_val = df.at[r, col]
        if isinstance(current_val, (int, float)):
            # Cari nilai numerik berikutnya di daftar positions
            next_numeric = None
            for j in range(index + 1, len(positions)):
                r_next, col_next = positions[j]
                candidate = df.at[r_next, col_next]
                if isinstance(candidate, (int, float)):
                    next_numeric = candidate
                    break
            # Jika ditemukan, hitung selisihnya
            if next_numeric is not None:
                usage_df.at[r, col] = next_numeric - current_val
            else:
                usage_df.at[r, col] = current_val
        else:
            # Jika bukan angka, salin nilai aslinya
            usage_df.at[r, col] = current_val

# Simpan hasil ke file baru
usage_df.to_excel("penggunaan_forklift.xlsx", index=False)

print("Perhitungan selesai. Data disimpan di 'penggunaan_forklift.xlsx'")
```
